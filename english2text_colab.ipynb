Upload an English audio file, transcribe it with Faster Whisper, and save the result inside an `output` folder.

## 2. Upload Audio

```python
from google.colab import files
from pathlib import Path

uploaded = files.upload()
audio_path = Path(next(iter(uploaded.keys())))
print(f"Uploaded: {audio_path}")
```

## 3. Transcribe

```python
from pathlib import Path
import torch
from faster_whisper import WhisperModel

# Options: "tiny", "base", "small", "medium", "large-v3"
# Start with "medium" for good accuracy/speed. Use "large-v3" for best accuracy.
MODEL_NAME = "medium"

device = "cuda" if torch.cuda.is_available() else "cpu"
compute_type = "float16" if device == "cuda" else "int8"

output_dir = Path("output")
output_dir.mkdir(exist_ok=True)
output_path = output_dir / f"{audio_path.stem}_{MODEL_NAME}_english.txt"

model = WhisperModel(MODEL_NAME, device=device, compute_type=compute_type)
segments, info = model.transcribe(str(audio_path), language="en", beam_size=5)

lines = []
for segment in segments:
    lines.append(segment.text.strip())

text = "\n".join(line for line in lines if line)
output_path.write_text(text, encoding="utf-8")

print(f"Detected language: {info.language}")
print(f"Saved transcription to: {output_path}")
```
