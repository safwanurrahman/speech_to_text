Upload a Bangla MP3 file, transcribe it with BanglaSpeech2Text, and save the result inside an `output` folder.

## 2. Upload MP3

```python
from google.colab import files
from pathlib import Path

uploaded = files.upload()
audio_path = Path(next(iter(uploaded.keys())))
print(f"Uploaded: {audio_path}")
```

## 3. Transcribe

```python
import os
from pathlib import Path
from banglaspeech2text import Speech2Text

# Options: "tiny", "base", "small", "large"
# For best accuracy, use "large". It downloads a big model and can take a while.
MODEL_NAME = "large"

# Helps avoid slow/stuck Hugging Face Xet downloads in some Colab sessions.
os.environ["HF_HUB_DISABLE_XET"] = "1"

output_dir = Path("output")
output_dir.mkdir(exist_ok=True)
output_path = output_dir / f"{audio_path.stem}_{MODEL_NAME}.txt"

stt = Speech2Text(MODEL_NAME)
text = stt.recognize(str(audio_path))

output_path.write_text(text, encoding="utf-8")
print(f"Saved transcription to: {output_path}")
```

## 4. Preview

```python
print(text[:2000])
```

## 5. Download Text File

```python
files.download(str(output_path))
```
